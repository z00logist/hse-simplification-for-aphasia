# Count Metaphors Index

### Hypothesis: we await that index will be falling from the original text to its most simplified form

Let's choose three random samples

```python
import pandas as pd

df = pd.read_csv("data/RuSimpleSentAphasia.csv")
random_samples = df.sample(n=10, random_state=42)

random_samples
```

```python
for index, row in random_samples.iterrows():
    print("Sample:", index)
    print("Source:", row['source'])
    print("Level 1:", row['level 1'])
    print("Level 2:", row['level 2'])
    print("Level 3:", row['level 3'])
    print("-" * 40)
```

```python
import pandas as pd

samples = [
    {
        "sample": 521,
        "texts": {
            "Source": "Почувствовав опасность, подопытные травы начали издавать запах, похожий на запах лаванды.",
            "Level 1": "При опасности подопытные травы начали издавать запах. Он был похож на запах лаванды.",
            "Level 2": "При опасности травы начали издавать запах. Он был похож на запах лаванды.",
            "Level 3": "При опасности травы издавали запах. Запах напоминал лаванду."
        },
        "metaphors": [
            {"dead": 2, "conventional": 0, "new": 2},
            {"dead": 1, "conventional": 0, "new": 1},
            {"dead": 1, "conventional": 0, "new": 1},
            {"dead": 1, "conventional": 0, "new": 1},
        ]
    },
    {
        "sample": 739,
        "texts": {
            "Source": "Именно в силу этой закоренелой привычки в вышеупомянутый первый понедельник апреля 1625 года горожане, услышав шум и не узрев ни желто – красных значков, ни ливрей слуг герцога Ришелье, устремились к гостинице «Вольный Мельник».",
            "Level 1": "Именно из-за этого в апреле 1625 года горожане побежали к гостинице 'Вольный мельник'. Они услышали шум, но не увидели ни желто-красные значки, ни слуг герцога Ришелье.",
            "Level 2": "Из-за этого в апреле 1625 года горожане побежали к гостинице 'Вольный мельник'. Они не увидели людей короля или Ришелье.",
            "Level 3": "Из-за этого в один день горожане побежали к гостинице 'Вольный мельник'. Они услышали шум. Но там не было людей короля или Ришелье."
        },
        "metaphors": [
            {"dead": 1, "conventional": 2, "new": 2},
            {"dead": 0, "conventional": 1, "new": 1},
            {"dead": 0, "conventional": 1, "new": 0},
            {"dead": 0, "conventional": 1, "new": 0},
        ]
    },
    {
        "sample": 823,
        "texts": {
            "Source": "Причастных к этому дипломатов будут выслать, невзирая на их должности, добавил он.",
            "Level 1": "Причастных к этому дипломатов любой должности будут высылать из страны.",
            "Level 2": "Причастных дипломатов будут выгонять из страны.",
            "Level 3": "Виновных дипломатов выгонят из страны."
        },
        "metaphors": [
            {"dead": 2, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 0, "conventional": 0, "new": 0},
            {"dead": 0, "conventional": 0, "new": 0},
        ]
    },
    {
        "sample": 662,
        "texts": {
            "Source": "В литре воды разведите ложку соли.",
            "Level 1": "В литре воды разведите ложку соли..",
            "Level 2": "В литре воды разведите ложку соли.",
            "Level 3": "В литре воды разведите ложку соли."
        },
        "metaphors": [
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
        ]
    },
    {
        "sample": 411,
        "texts": {
            "Source": "Поводом для его сооружения стал трехсотлетний юбилей Нижегородской губернии, созданной царским указом в 1714 году.",
            "Level 1": "Памятник открыли в честь юбилея Нижегородской области. Ее основали в 1714 году.",
            "Level 2": "Памятник открыли в честь юбилея Нижегородской области.",
            "Level 3": "Памятник открыли в честь юбилея области."
        },
        "metaphors": [
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
        ]
    },
    {
        "sample": 680,
        "texts": {
            "Source": "Тогда стоит задуматься о бумажных аналогах – это гигиенично и практично.",
            "Level 1": "Тогда стоит задуматься о бумажных аналогах. Это гигиенично и практично.",
            "Level 2": "Стоит задуматься о бумажных вариантах. Это гигиенично и практично.",
            "Level 3": "Тогда стоит выбрать бумажные. Это гигиенично и удобно."
        },
        "metaphors": [
            {"dead": 0, "conventional": 0, "new": 1},
            {"dead": 0, "conventional": 0, "new": 1},
            {"dead": 0, "conventional": 0, "new": 1},
            {"dead": 0, "conventional": 0, "new": 0},
        ]
    },
    {
        "sample": 628,
        "texts": {
            "Source": "По утверждению BBC, вспышка похожего заболевания произошла в 1998 году в Тайване.",
            "Level 1": "Вспышка похожего заболевания произошла в 1998 году в Тайване.",
            "Level 2": "Вспышка похожего заболевания была в 1998 году в Тайване.",
            "Level 3": "Похожая вспышка была в 1998 году в Тайване."
        },
        "metaphors": [
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
            {"dead": 1, "conventional": 0, "new": 0},
        ]
    },
    {
        "sample": 513,
        "texts": {
            "Source": "В нем планировалось замерить ту мощность, которую выдают объединенные компьютеры.",
            "Level 1": "Организаторы планировали замерить мощность объединенных компьютеров.",
            "Level 2": "Организаторы хотели измерить мощность союза компьютеров.",
            "Level 3": "Организаторы измерили мощность союза компьютеров."
        },
        "metaphors": [
            {"dead": 0, "conventional": 1, "new": 0},
            {"dead": 0, "conventional": 1, "new": 0},
            {"dead": 0, "conventional": 0, "new": 1},
            {"dead": 0, "conventional": 0, "new": 1},
        ]
    },
    {
        "sample": 774,
        "texts": {
            "Source": "Умелые и сноровистые, хоббиты, однако, терпеть не могли – да не могут и поныне – устройств сложнее кузнечных мехов, водяной мельницы и прялки.",
            "Level 1": "Но умелые хоббиты терпеть не могут устройств сложнее кузнечных мехов, водяной мельницы и прялки.",
            "Level 2": "Однако хоббиты не любят устройства сложнее кузнечных мехов и мельницы.",
            "Level 3": "Однако хоббиты не любят сложные устройства."
        },
        "metaphors": [
            {"dead": 0, "conventional": 0, "new": 1},
            {"dead": 0, "conventional": 0, "new": 1},
            {"dead": 0, "conventional": 0, "new": 1},
            {"dead": 0, "conventional": 0, "new": 0},
        ]
    },
    {
        "sample": 136,
        "texts": {
            "Source": "Далее в прогнозе ожидается небольшой перерыв и понижение температуры до плюс одного – минус одного.",
            "Level 1": "Затем на улице похолодает.",
            "Level 2": "Затем на улице похолодает.",
            "Level 3": "Затем на улице похолодает."
        },
        "metaphors": [
            {"dead": 0, "conventional": 1, "new": 0},
            {"dead": 0, "conventional": 0, "new": 0},
            {"dead": 0, "conventional": 0, "new": 0},
            {"dead": 0, "conventional": 0, "new": 0},
        ]
    },
]


def calculate_MFI(dead: int, dormant: int, live: int, total_metaphors: int) -> int:
    return ((live * 3) + (dormant * 2) + dead) / total_metaphors if total_metaphors > 0 else 0


def calculate_MII(total_metaphors: int, total_words: int) -> int:
    return total_metaphors / total_words * 100 if total_words > 0 else 0


results = []
for sample in samples:
    for level, text in sample["texts"].items():
        metaphors = sample["metaphors"][list(sample["texts"].keys()).index(level)]
        w, n, s = metaphors["dead"], metaphors["conventional"], metaphors["new"]
        t = w + n + s
        Q = len(text.split())
        
        C = (1 * w + 1.5 * n + 3 * s) / t if t > 0 else 0
        F = t / Q if Q > 0 else 0
        
        live = s
        dormant = n
        dead = w

        MFI = calculate_MFI(dead, dormant, live, t)
        MII = calculate_MII(t, Q)
        
        results.append({
            "Sample": sample["sample"],
            "Level": level,
            "Words": Q,
            "Metaphors Total": t,
            "Creativity Index (C)": round(C, 3),
            "Frequency Index (F)": round(F, 3),
            "MFI": round(MFI, 3),
            "MII": round(MII, 3),
        })

results_df = pd.DataFrame(results)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

average_indices = results_df.groupby("Level")[["Creativity Index (C)", "Frequency Index (F)"]].mean().reset_index()
order = ["Source", "Level 1", "Level 2", "Level 3"]
average_indices["Level"] = pd.Categorical(average_indices["Level"], categories=order, ordered=True)
average_indices = average_indices.sort_values("Level")

plt.figure(figsize=(10, 6))
sns.lineplot(data=average_indices, x="Level", y="Creativity Index (C)", marker="o", label="Creativity Index (C)")
sns.lineplot(data=average_indices, x="Level", y="Frequency Index (F)", marker="o", label="Frequency Index (F)")

plt.title("Average Metaphor Indices by Level", fontsize=16)
plt.xlabel("Level", fontsize=14)
plt.ylabel("Index Value", fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title="Index Type", fontsize=12, title_fontsize=14)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.lineplot(data=average_indices, x="Level", y="MFI", marker="o", label="Metaphor Frequency Index (MFI)")
sns.lineplot(data=average_indices, x="Level", y="MII", marker="o", label="Metaphor Intensity Index (MII)")

plt.title("Average Metaphor Frequency and Intensity Indices by Level", fontsize=16)
plt.xlabel("Level", fontsize=14)
plt.ylabel("Index Value", fontsize=14)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.legend(title="Index Type", fontsize=12, title_fontsize=14)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
